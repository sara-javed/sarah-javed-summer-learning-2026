#**Section 4: Data Visualization (Matplotlib & Seaborn)**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for better looking plots
sns.set_style("whitegrid")
```

## **Wine Quality Distribution (Count Plot)**

This chart shows the number of wines available for each quality score.

```python
plt.figure(figsize=(4,3))
sns.countplot(data=df, x="quality")

plt.title("Distribution of Wine Quality")
plt.xlabel("Quality Score")
plt.ylabel("Number of Wines")

plt.show()
```

### Observation

- Most wines have quality scores of **5** and **6**.
- Very few wines have quality scores of **3** and **8**.
- The dataset is imbalanced because some quality scores have significantly fewer samples than others.
- This imbalance should be considered before building a machine learning model.

##  **Distribution of Alcohol Content**

This histogram shows how the alcohol content is distributed across the wine samples.

```python
plt.figure(figsize=(4,3))

sns.histplot(data=df, x="alcohol", bins=20, kde=True)

plt.title("Distribution of Alcohol Content")
plt.xlabel("Alcohol (%)")
plt.ylabel("Frequency")

plt.show()
```

### Observations: Alcohol Content Distribution

- Most wine samples have an alcohol content between **9% and 12%**.
- The highest concentration of wines is around **9.5% to 10%** alcohol.
- Wines with alcohol content above **12%** are less common.
- The alcohol content is not evenly distributed across the dataset.

##  **Correlation Heatmap**

The correlation heatmap visualizes the relationships between numerical features in the dataset. Correlation values range from -1 to +1, where positive values indicate a positive relationship and negative values indicate a negative relationship.

```python
plt.figure(figsize=(12, 8))

sns.heatmap(df.corr(), annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Wine Quality Dataset")

plt.show()
```

### Observations: Correlation Heatmap

- Alcohol has the strongest positive correlation with wine quality (**0.48**).
- Volatile acidity has the strongest negative correlation with wine quality (**-0.40**).
- Strong correlations are also observed between fixed acidity & pH, fixed acidity & density, and free & total sulfur dioxide.
- The heatmap highlights key relationships between features and wine quality.

##  **Alcohol Content vs Wine Quality**

This box plot compares the distribution of alcohol content across different wine quality scores.

```python
plt.figure(figsize=(4,3))

sns.boxplot(data=df, x="quality", y="alcohol")

plt.title("Alcohol Content by Wine Quality")
plt.xlabel("Quality Score")
plt.ylabel("Alcohol (%)")

plt.show()
```

### Observations: Alcohol Content vs. Wine Quality

- Higher-quality wines generally have higher alcohol content.
- The median alcohol level increases as the quality score increases.
- Some quality groups contain outliers with unusually high alcohol content.
- The box plot supports the positive correlation between alcohol and wine quality.

## **Relationship Between Alcohol Content and Wine Quality**

This scatter plot visualizes the relationship between alcohol content and wine quality.

```python
plt.figure(figsize=(4,3))

sns.scatterplot(data=df, x="alcohol", y="quality")

plt.title("Alcohol vs Wine Quality")
plt.xlabel("Alcohol (%)")
plt.ylabel("Quality Score")

plt.show()
```

## **Pair Plot of Selected Features**

This pair plot visualizes the relationships and distributions of selected numerical features.

```python
selected_features = ["alcohol", "volatile acidity", "citric acid", "quality"]

sns.pairplot(df[selected_features], hue="quality")

plt.show()
```

### Observations: Pair Plot

- Alcohol shows a positive relationship with wine quality.
- Volatile acidity tends to decrease as wine quality increases.
- Most features show overlapping distributions, while a few exhibit clear trends with quality.
- The pair plot provides an overview of relationships among selected features.

##  **Distribution of Alcohol Content by Wine Quality**

This violin plot shows the distribution and density of alcohol content across different wine quality scores.

```python
plt.figure(figsize=(4,3))

sns.violinplot(data=df, x="quality", y="alcohol")

plt.title("Alcohol Distribution by Wine Quality")
plt.xlabel("Quality Score")
plt.ylabel("Alcohol (%)")

plt.show()
```

### Observations: Violin Plot

- Higher-quality wines generally have higher alcohol content.
- The alcohol distribution shifts upward as the quality score increases.
- The distribution of alcohol varies across different quality levels.
- The violin plot provides a clear view of both the distribution and density of alcohol content.
