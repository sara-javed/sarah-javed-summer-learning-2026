# **Calculus Introduction | Why Calculus Exists**

**Calculus** is a branch of mathematics that helps us study change and accumulation.

It mainly answers two questions:

1. How fast is something changing? (**Derivative**)
2. How much has accumulated? (**Integral**)

**Why was Calculus invented?**

Many real-world problems cannot be solved using simple geometry.

Examples:
- Area under a curve
- Speed of a moving car
- Population growth
- Machine Learning optimization

Calculus provides mathematical tools to solve these problems.

• Break a difficult shape into many tiny pieces.

• Solve each small piece.

• Add all the pieces together.

**Two Main Parts of Calculus**

**Derivative:**	  Measures how fast something changes

**Integral:**	    Measures total accumulation

**Real-Life Applications**

* Physics:
Distance from speed

* Economics:
Total profit

* Machine Learning:
Gradient Descent

* Engineering:
Work Done

* Biology:
Population Growth


**Python Visualization**

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 5, 100)
y = x**2

plt.plot(x, y)
plt.fill_between(x, y, alpha=0.3)

plt.title("Area Under the Curve")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)

plt.show()
```
