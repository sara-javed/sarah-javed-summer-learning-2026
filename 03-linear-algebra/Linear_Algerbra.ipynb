# **2️. Dot Product**

The **Dot Product** is obtained by multiplying the corresponding elements of two vectors and then adding the results.

The output of a Dot Product is always a scalar (single number).

Formula

A · B = a₁b₁ + a₂b₂ + ... + aₙbₙ

##  Intuition

The Dot Product measures how similar two vectors are in direction.

If two vectors point in the same direction,

➡ Dot Product is Positive

If they are perpendicular,

➡ Dot Product is Zero

If they point in opposite directions,

➡ Dot Product is Negative

##  Example 1

A = [2,3]

B = [4,5]

Step 1

2 × 4 = 8

3 × 5 = 15

Step 2

8 + 15 = 23

Answer = 23

##   Example 2

A = [1,2,3]

B = [4,5,6]

Calculation

1×4 = 4

2×5 = 10

3×6 = 18

Total = 32

```python
import numpy as np

A = np.array([2,3])

B = np.array([4,5])

dot = np.dot(A,B)

print(dot)
```

# **6️. Systems of Linear Equations**



A **system of linear equations** is a group of two or more linear equations that share the same variables.

The goal is to find values of the variables that satisfy all equations at the same time.

##  Simple Explanation

Consider the following equations.

**x + y = 8**

**2x + y = 10**

We need to find values of **x** and **y** that satisfy both equations.

The common solution is the answer to the system.

##  Example 1

x + y = 6

x - y = 2

Solution

x = 4

y = 2

```python
import numpy as np

A = np.array([
    [1,1],
    [2,1]
])

B = np.array([8,10])

solution = np.linalg.solve(A,B)

print(solution)
```


**Systems of linear equations** are used in:

- Linear Regression

- Scientific Computing

# **7️. Gaussian Elimination**



**Gaussian Elimination** is a systematic method used to solve a system of linear equations.

It works by performing row operations to simplify a matrix until the solution can be found easily.

Instead of solving equations directly, we first convert them into a matrix and then simplify that matrix step by step.

##  Simple Explanation

Imagine you have two equations:

x + y = 8

2x + y = 10

Instead of solving them manually, we arrange the equations into a matrix and perform simple row operations.


The goal is to make the matrix easier to solve.



Only three row operations are allowed:

1. Swap two rows.
2. Multiply a row by a non-zero number.
3. Add or subtract one row from another.

These operations do not change the final answer.

##  Example 1

System of Equations

x + y = 8

2x + y = 10

Subtract Row 1 from Row 2.

New equation:

x = 2

Substitute x = 2 into the first equation.

2 + y = 8

y = 6

Final Solution

x = 2

y = 6

```python
import numpy as np

A = np.array([[1,1],
              [2,1]])

B = np.array([8,10])

solution = np.linalg.solve(A,B)

print(solution)
```



**Gaussian Elimination** is used in:

- Solving linear equations
- Linear Regression
- Matrix Inverse calculations
- Engineering problems
- Scientific Computing

Most libraries such as NumPy perform these operations internally.

# **8️. Matrix Inverse**



The **inverse** of a square matrix is another matrix that reverses the effect of the original matrix.

If A is a matrix and A⁻¹ is its inverse, then

A × A⁻¹ = I

where I is the Identity Matrix.

##  Simple Explanation

Think about ordinary numbers.

5 × (1/5) = 1

Similarly,

A × A⁻¹ = I

The Identity Matrix behaves like the number 1 in matrix multiplication.

If a matrix has an inverse, solving matrix equations becomes much easier.

##  Example 1

Matrix

A

[[1,2],

    [3,4]]

Its inverse is

[[-2,1],
  
    [1.5,-0.5]]

Multiplying A by its inverse gives the Identity Matrix.

```python
import numpy as np

A = np.array([[1,2],
              [3,4]])

inverse = np.linalg.inv(A)

print(inverse)
```



**Matrix Inverse** is used in:

- Scientific Computing

# **10. Matrix Rank**



The **rank of a matrix** is the maximum number of linearly independent rows or columns in the matrix.

In simple words, the rank tells us how much unique information a matrix contains.

If some rows or columns are copies (or combinations) of others, they do not increase the rank.

##  Simple Explanation

Imagine you have a classroom with three students.

Student 1 marks:

[80, 75]

Student 2 marks:

[90, 85]

Student 3 marks:

[160, 150]

Notice that Student 3's marks are exactly twice Student 1's marks.

This means Student 3 does not provide any new information.

Therefore, the matrix contains redundant information.

The rank tells us the number of independent (unique) rows or columns.

##  Example 1

Matrix

[

 [1,2],

 [2,4]

]

The second row is exactly twice the first row.

No new information is added.

Rank = 1

##  Example 2

Suppose we have a dataset.

| Age | Salary | Income |
|------|---------|---------|
|20|50000|100000|
|25|70000|140000|
|30|90000|180000|

Here,

Income = Salary × 2

The Income column does not add new information because it is completely dependent on the Salary column.

Therefore, the feature matrix has redundant information, and its rank is lower than the total number of columns.

```python
import numpy as np

A = np.array([
    [1,2],
    [2,4]
])

print("Rank =", np.linalg.matrix_rank(A))
```



**Matrix Rank** is important in Data Science because it helps us identify redundant features.

Applications include:

- Detecting multicollinearity
- Feature selection
- Principal Component Analysis (PCA)
- Singular Value Decomposition (SVD)
- Recommendation Systems
- Dimensionality Reduction

# **11. Vector Spaces**



A **Vector Space** is a collection of vectors where two operations are defined:

- **Vector Addition**
- **Scalar Multiplication**

These operations always produce another vector that belongs to the same vector space.

In simple words, a Vector Space is a mathematical environment where vectors follow specific rules and remain inside the same space after addition or scalar multiplication.

#  Simple Explanation

Imagine you have a box of LEGO blocks.

You can:

- Join two LEGO blocks together.
- Make a LEGO block bigger or smaller.

No matter what you do, you still have LEGO blocks.

Similarly, in a **Vector Space**, when we add vectors or multiply them by a number (scalar), the result is always another vector in the same space.

This is why it is called a **Vector Space**.

#  Example 1

Let

\[
A = (2,1)
\]

\[
B = (4,3)
\]

### Vector Addition

\[
A+B=(6,4)
\]

The result is still a vector.

### Scalar Multiplication

\[
3A=(6,3)
\]

Again, the result is still a vector.

Therefore, both operations remain inside the same Vector Space.

```python
import numpy as np

A = np.array([2, 1])
B = np.array([4, 3])

print("Vector A =", A)
print("Vector B =", B)

print("\nA + B =", A + B)

print("3 × A =", 3 * A)
```


**Vector Spaces** are the mathematical foundation of many Data Science and Machine Learning algorithms.

Applications include:

#  **12. Span, Basis & Dimension**



**Span, Basis, and Dimension** are three closely related concepts in Linear Algebra.

- **Span** tells us all the vectors that can be created using a given set of vectors.
- **Basis** is the smallest set of vectors that can generate (span) the entire vector space.
- **Dimension** is simply the number of vectors in the basis.

These concepts are fundamental in Machine Learning, Data Science, and Dimensionality Reduction techniques such as PCA.

#  What is Span?

The **span** of a set of vectors is the collection of **all possible vectors** that can be created by adding the vectors together and multiplying them by scalars.

In simple words,

**Span tells us how much space the given vectors can cover.**

#  Simple Explanation

Imagine you have only two paint colors:

- 🟡 Yellow
- 🔵 Blue

By mixing them in different amounts, you can create many different shades of green.

All the colors you can produce represent the **span** of the two original colors.

Similarly, vectors can be combined to create many new vector

#  Example 1 (Span)

Let

\[
A=(1,0)
\]

Multiplying the vector by different numbers gives

\[
2A=(2,0)
\]

\[
5A=(5,0)
\]

\[
-3A=(-3,0)
\]

All these vectors lie on the **x-axis**.

Therefore,

**Span(A) = x-axis**

```python
import numpy as np

A = np.array([1, 0])

print("A =", A)

print("2A =", 2 * A)
print("5A =", 5 * A)
print("-3A =", -3 * A)
```

#  What is Basis?

A **Basis** is the **smallest set of linearly independent vectors** that can generate the entire vector space.

In simple words,

A Basis contains only the **essential vectors**.

It does **not** contain duplicate or redundant vectors.

#  Simple Explanation

Imagine you are building a tent.

Only **two poles** are enough to support it.

Adding five extra poles does not make the tent more complete.

Those extra poles are **redundant**.

Similarly,

A Basis contains only the minimum vectors needed to generate the entire space.

#  Basis Example

For a 2D plane,

\[
(1,0)
\]

and

\[
(0,1)
\]

are enough to generate every point.

Therefore,

These two vectors form a **Basis** of the 2D plane.

#  What is Dimension?

The **Dimension** of a vector space is the **number of vectors in its Basis**.

In simple words,

Dimension tells us the **minimum number of independent directions** needed to describe the entire space.

#  Dimension Examples

### Line

One basis vector

Dimension = 1

---

### Plane

Two basis vectors

Dimension = 2

---

### 3D Space

Three basis vectors

Dimension = 3

```python
import numpy as np

A = np.array([1, 0])
B = np.array([0, 1])

basis = np.column_stack((A, B))

print("Basis Matrix:\n")
print(basis)

print("\nRank =", np.linalg.matrix_rank(basis))
print("Dimension =", basis.shape[1])
```

#  **13. Linear Independence**

A set of vectors is **linearly independent** if **none of the vectors can be written as a combination of the others**.

In simple words, each vector provides **new and unique information**.

If one vector can be created from the others, then the vectors are **linearly dependent**.

#  Simple Explanation

Imagine three students are doing an assignment.

- Student A writes the assignment by himself.
- Student B also writes his own assignment.
- Student C copies Student A's assignment.

Student C does not contribute any new work.

Similarly, if one vector is just a copy (or multiple) of another vector, it is **linearly dependent**.

Only vectors that provide **new information** are called **linearly independent**.

#  Example 1 (Independent Vectors)

Consider the vectors

\[
A=(1,0)
\]

\[
B=(0,1)
\]

Neither vector can be created from the other.

Therefore,

**A and B are Linearly Independent.**

```python
import numpy as np

A = np.array([1, 0])
B = np.array([0, 1])

print("Vector A =", A)
print("Vector B =", B)

print("\nThese vectors are linearly independent because neither can be obtained from the other.")
```

#  **14. Orthogonality**

Two vectors are **orthogonal** if their **dot product is equal to zero**.

Geometrically, orthogonal vectors are **perpendicular (90°)** to each other.

In simple words, orthogonal vectors point in completely different directions and do not share overlapping information.

#  Simple Explanation

Imagine two roads crossing each other at a perfect **90° angle**.

One road goes from **East to West**.

The other road goes from **North to South**.

These roads are perpendicular to each other.

Similarly, two vectors are called **orthogonal** when they meet at a **90° angle**.

#  Example 1 (Orthogonal Vectors)

Let

\[
A=(1,0)
\]

\[
B=(0,1)
\]

Dot Product

\[
(1\times0)+(0\times1)=0
\]

Since the dot product is **0**, the vectors are **Orthogonal**.

```python
import numpy as np

A = np.array([1, 0])
B = np.array([0, 1])

dot_product = np.dot(A, B)

print("Vector A =", A)
print("Vector B =", B)
print("Dot Product =", dot_product)

if dot_product == 0:
    print("The vectors are Orthogonal.")
else:
    print("The vectors are NOT Orthogonal.")
```

#  **15. Projections**



A **Projection** of a vector is the closest point of that vector on another vector or subspace.

In simple words, a projection is like the **shadow** of one vector on another vector.

Projection helps us find the useful part of a vector in a particular direction.

#  Simple Explanation

Imagine a tree standing in sunlight.

The tree creates a shadow on the ground.

-  Tree = Original Vector
-  Shadow = Projection

Similarly, a vector can be projected onto another vector to find its component in that direction.

#  Example 1

Suppose

\[
A=(4,3)
\]

and

\[
B=(1,0)
\]

Since vector **B** lies along the x-axis, projecting **A** onto **B** keeps only the x-component.

Projection of **A** onto **B**

\[
(4,0)
\]

The y-component is removed.

```python
import numpy as np

A = np.array([4, 3])
B = np.array([1, 0])

projection = np.dot(A, B) / np.dot(B, B) * B

print("Vector A =", A)
print("Vector B =", B)
print("Projection of A onto B =", projection)
```

#  **17. Cosine Similarity**



**Cosine Similarity** measures how similar two vectors are based on the **angle between them**, regardless of their magnitude.

Its value ranges from:

- **1** → Exactly the same direction
- **0** → Perpendicular (Orthogonal)
- **-1** → Opposite direction

Cosine Similarity is widely used to compare documents, images, and embeddings.

#  Simple Explanation

Imagine two people walking.

- If both walk in the **same direction**, they are very similar.
- If they walk at **90°**, they are unrelated.
- If they walk in **opposite directions**, they are completely different.

Cosine Similarity measures the **direction**, not the distance.

#  Formula

\[
\text{Cosine Similarity}=
\frac{A\cdot B}
{||A||\times||B||}
\]

where

- \(A \cdot B\) = Dot Product
- \(||A||\) = Norm of Vector A
- \(||B||\) = Norm of Vector B

#  Example

Let

\[
A=(1,2)
\]

\[
B=(2,4)
\]

These vectors point in the same direction.

Therefore,

**Cosine Similarity = 1**

```python
import numpy as np

A = np.array([1,2])
B = np.array([2,4])

cosine = np.dot(A,B)/(np.linalg.norm(A)*np.linalg.norm(B))

print("Cosine Similarity =", cosine)
```

#  **19. Eigenvalues & Eigenvectors**





An **Eigenvector** is a special vector whose **direction does not change** after a linear transformation.

An **Eigenvalue** tells us how much that eigenvector is stretched or compressed.

#  Simple Explanation

Imagine an arrow drawn on a rubber sheet.

When the sheet is stretched,

- Most arrows change both **length** and **direction**.
- Some special arrows only change **length**.

These special arrows are **Eigenvectors**.

The stretching factor is called the **Eigenvalue**.

#  Formula

\[
Av=\lambda v
\]

where

- \(A\) = Matrix
- \(v\) = Eigenvector
- \(\lambda\) = Eigenvalue

The vector keeps the same direction and only its magnitude changes.

#  Example

Consider

\[
A=
\begin{bmatrix}
2&0\\
0&3
\end{bmatrix}
\]

The vector

\[
(1,0)
\]

remains in the same direction after transformation.

It is an **Eigenvector** with **Eigenvalue = 2**.

```python
import numpy as np

A = np.array([[2,0],
              [0,3]])

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:\n", eigenvalues)

print("\nEigenvectors:\n")
print(eigenvectors)
```

#  **20. Positive Definite Matrices**



A **Positive Definite Matrix** is a symmetric matrix whose **all eigenvalues are positive**.

Mathematically,

\[
x^T A x > 0
\]

for every non-zero vector **x**.

In simple words, a Positive Definite Matrix always produces a positive value and represents a stable mathematical system.

#  Simple Explanation

Imagine a smooth bowl.

If you place a ball anywhere inside it, the ball always rolls toward the **single lowest point**.

A Positive Definite Matrix behaves like this bowl.

It always has one stable minimum.

#  Example

Consider

\[
A=
\begin{bmatrix}
2&0\\
0&3
\end{bmatrix}
\]

Eigenvalues are

2 and 3

Since both are positive,

the matrix is **Positive Definite**.

```python
import numpy as np

A = np.array([[2,0],
              [0,3]])

eigenvalues = np.linalg.eigvals(A)

print("Eigenvalues:", eigenvalues)

print("Positive Definite:",
      np.all(eigenvalues > 0))
```

#  **21. Singular Value Decomposition (SVD)**



Singular Value Decomposition (SVD) is a matrix factorization technique.

It decomposes any matrix into three matrices.

\[
A=U\Sigma V^T
\]

where

- **U** → Left Singular Vectors
- **Σ** → Singular Values
- **Vᵀ** → Right Singular Vectors

#  Simple Explanation

Imagine breaking a large machine into three smaller parts.

Instead of working with one complicated machine,

we work with three simpler parts.

SVD does the same thing for matrices.

#  Example

A matrix

\[
A=
\begin{bmatrix}
1&2\\
3&4
\end{bmatrix}
\]

can be decomposed into

\[
U\Sigma V^T
\]

This decomposition makes calculations easier.

```python
import numpy as np

A = np.array([[1,2],
              [3,4]])

U,S,Vt = np.linalg.svd(A)

print("U=\n",U)

print("\nSingular Values=\n",S)

print("\nVᵀ=\n",Vt)
```

#  **22. Principal Component Analysis (PCA)**



Principal Component Analysis (PCA) is a dimensionality reduction technique.

It transforms the original features into a new set of **orthogonal principal components**.

The first principal component captures the maximum variance, the second captures the next highest variance, and so on.

#  Simple Explanation

Imagine you have **100 features** in a dataset.

Many of them contain similar information.

PCA combines these features into a smaller number of important features while preserving most of the useful information.

For example,

100 Features

↓

10 Principal Components

# Example

Suppose a dataset has

- Age
- Salary
- Income

Some features are highly correlated.

PCA creates new features called **Principal Components** that contain most of the important information while removing redundancy.

```python
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris

iris = load_iris()

X = iris.data

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

print("Original Shape:", X.shape)

print("Reduced Shape:", X_pca.shape)

print("Explained Variance Ratio:")

print(pca.explained_variance_ratio_)
```
